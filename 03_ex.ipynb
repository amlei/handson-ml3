## 导入库

## 加载数据

```python
train_data = pd.read_csv("datasets/titanic/train.csv").set_
test_data = pd.read_csv("datasets/titanic/test.csv")

y = train_data["Survived"]
y
```

## 幸存率

```python
men = train_data.loc[train_data.Sex == 'male']["Survived"]
rate_men = sum(men) / len(men)
rate_men
```

```python
women = train_data.loc[train_data.Sex == 'female']["Survived"]
rate_women = sum(women) / len(women)
rate_women
```

## 训练模型

```python
y = train_data["Survived"]
features = ["Pclass", "Sex", "SibSp", "Parch"]
X_train = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])
```

```python
X_train
```

```python
X_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier


model = RandomForestClassifier()
model.fit(X_train, y)
```

```python
perdiciton = model.predict(X_test)
```

```python
perdiciton.shape
```

## 数据导出

```python
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': perdiciton})
output.to_csv("./datasets/titanic/output.csv", index=False)
```
